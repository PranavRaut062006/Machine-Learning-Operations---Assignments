```python
import pandas as pd

students = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Aarav", "Vivaan", "Aditya", "Arjun", "Rohan",
        "Ishaan", "Karan", "Rahul", "Ananya", "Priya"
    ],
    "Age": [20, 21, 20, 22, 21, 20, 22, 21, 20, 22],
    "Branch": [
        "CSE", "IT", "ECE", "CSE", "Mechanical",
        "CSE", "IT", "ECE", "CSE", "IT"
    ],
    "CGPA": [8.5, 9.1, 8.2, 9.3, 7.8, 8.9, 8.1, 8.7, 9.0, 8.4]
}

df = pd.DataFrame(students)

df.to_csv("../data/students.csv", index=False)

df
```

```python
new_students = pd.DataFrame({
   "Student_ID": [111, 112, 113],
   "Name": ["Sneha", "Vikram", "Neha"],
   "Age": [21, 22, 20],
   "Branch": ["CSE", "Mechanical", "ECE"],
   "CGPA": [8.6, 7.9, 9.2]
})

df = pd.concat([df, new_students], ignore_index=True)

df.to_csv("../data/students.csv", index=False)

df
```
