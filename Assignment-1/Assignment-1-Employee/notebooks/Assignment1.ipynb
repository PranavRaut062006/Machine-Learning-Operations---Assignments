```python
import pandas as pd

employees = {
   "Employee_ID": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
   "Employee_Name": [
      "Amit", "Neha", "Rahul", "Priya", "Rohan",
      "Sneha", "Karan", "Anjali", "Vikram", "Pooja"
   ],
   "Department": [
      "HR", "IT", "Finance", "Marketing", "Sales",
      "Operations", "IT", "Finance", "HR", "Sales"
   ],
   "Salary": [
      35000, 45000, 50000, 40000, 38000,
      42000, 55000, 48000, 36000, 41000
   ]
}

df = pd.DataFrame(employees)

df.to_csv("../data/employees.csv", index=False)

df
```

```python
print("Number of employees:", len(df))
print("Dataset shape:", df.shape)
```

```python
df["Salary"] = df["Salary"] * 1.10
df.to_csv("../data/employees.csv", index=False)

df
```

```python
print("Number of employees:", len(df))
print("Average salary:", df["Salary"].mean())
```

```python
df["Years_of_Experience"] = [2, 4, 5, 3, 2, 6, 7, 5, 3, 4]

df.to_csv("../data/employees.csv", index=False)

df
```

```python
print("Columns:", list(df.columns))
print("Dataset shape:", df.shape)
```

```python
df["Performance_Rating"] = [
    "Excellent",
    "Very Good",
    "Good",
    "Excellent",
    "Average",
    "Very Good",
    "Excellent",
    "Good",
    "Average",
    "Very Good"
]

df.to_csv("../data/employees.csv", index=False)

df
```

```python
print("Columns:", list(df.columns))
print("Dataset shape:", df.shape)
```

```python
new_employees = pd.DataFrame({
    "Employee_ID": [211, 212, 213],
    "Employee_Name": ["Arjun", "Meera", "Sahil"],
    "Department": ["IT", "Marketing", "Operations"],
    "Salary": [52000, 46000, 43000],
    "Years_of_Experience": [5, 3, 4],
    "Performance_Rating": ["Excellent", "Good", "Very Good"]
})

df = pd.concat([df, new_employees], ignore_index=True)

df.to_csv("../data/employees.csv", index=False)

df
```

```python
print("Number of employees:", len(df))
print("Dataset shape:", df.shape)
```

```python
department_mapping = {
    "HR": "Human Resources",
    "IT": "Engineering",
    "Finance": "Accounts & Finance",
    "Marketing": "Business Development",
    "Sales": "Sales & Marketing",
    "Operations": "Business Operations"
}

df["Department"] = df["Department"].replace(department_mapping)

df.to_csv("../data/employees.csv", index=False)

df
```
